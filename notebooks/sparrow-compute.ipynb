# Sparrow Compute: lazy expressions over sparrow arrays

This notebook runs entirely in your browser: [xeus-cpp](https://github.com/compiler-research/xeus-cpp) interprets C++20 compiled to WebAssembly,
with [sparrow](https://github.com/sparrow-org/sparrow) and the header-only [sparrow-compute](https://github.com/sparrow-org/sparrow-compute) pre-installed.

Lazy [xtensor](https://github.com/xtensor-stack/xtensor) expressions over `sp::primitive_array`; `eval` materializes a new Sparrow array.

The demo notebooks: [sparrow.ipynb](sparrow.ipynb) (arrays), [sparrow-ipc.ipynb](sparrow-ipc.ipynb) (Arrow IPC with pyarrow), [sparrow-compute.ipynb](sparrow-compute.ipynb) (lazy compute) and [rockfinch.ipynb](rockfinch.ipynb) (C++ extension module for Python).

Run the cells from top to bottom (`Shift+Enter`). A C++ name can only be defined once per kernel: restart the kernel before re-running the notebook.

```c++
#pragma clang diagnostic ignored "-W#warnings"
// Must match the compile definitions of the WebAssembly sparrow package
#define SPARROW_USE_LARGE_INT_PLACEHOLDERS
#include <format>
#include <iostream>
#include <string>
#include <vector>

#include <sparrow.hpp>
#include <sparrow/config/sparrow_version.hpp>
#include <sparrow-compute.hpp>

namespace sp = sparrow;

std::cout << "sparrow " << sp::SPARROW_VERSION_MAJOR << "." << sp::SPARROW_VERSION_MINOR << "." << sp::SPARROW_VERSION_PATCH << std::endl;
std::cout << "sparrow-compute " << sp::compute::SPARROW_COMPUTE_VERSION_MAJOR << "." << sp::compute::SPARROW_COMPUTE_VERSION_MINOR << "." << sp::compute::SPARROW_COMPUTE_VERSION_PATCH << std::endl;
```

```c++
sp::primitive_array<double> a{1.0, 2.0, 3.0};
sp::primitive_array<double> b{4.0, 5.0, 6.0};
auto result = sp::compute::eval(sp::compute::sqrt((a + b) / 2.0));
std::cout << std::format("{}", result) << std::endl;
```

```c++
auto lazy = a * b;  // nothing computed yet
std::cout << std::format("{}", sp::compute::eval(lazy)) << std::endl;
std::cout << std::format("{}", sp::compute::eval(sp::compute::sin(sp::compute::as_expression(a)))) << std::endl;
```
